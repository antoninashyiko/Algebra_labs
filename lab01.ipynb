б) напишіть цикл, що виводить числа від 1 до 10; <br>

```sage
n=float(input('enter number: '))
if n>0:
    print("positive")
elif n<0:
    print('negative')
else:
    print('zero')
```

```sage
for n in range(1, 11):
    print(n)
```

```sage
def func(n):
    flag=True
    for i in range(2, int(sqrt(n)+1)):
        if n%i==0:
            flag=False
            break
        else:
            continue
    return flag
n=int(input('enter number: '))
if func(n):
    print('prime')
else:
    print('not prime')
```

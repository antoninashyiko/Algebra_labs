### 1. Структура абелевих груп

Розглянемо абелеві групи порядку $400$:
$$
\mathbb{Z}_{400}, \quad \mathbb{Z}_{16}\times\mathbb{Z}_{25}, \quad \mathbb{Z}_{20}\times\mathbb{Z}_{20}, \quad \mathbb{Z}_{4}\times\mathbb{Z}_{100}, \quad \mathbb{Z}_{2}\times\mathbb{Z}_{10}\times\mathbb{Z}_{20}, \quad \mathbb{Z}_{505}^{*}.
$$
Для кожної з цих груп виконайте наступні кроки:
1. Знайдіть елементарні дільники групи.
2. Розкладіть групу у добуток циклічних $p$-груп.
3. Визначте кількість елементів кожного можливого порядку.
4. Порівняйте групи та визначте, які з них є ізоморфними.


```sage
Z400 = CyclicPermutationGroup(400) # введення всіх груп
Z400 = Z400. as_AbelianGroup()
Z16x25 = AbelianGroup ( [16 , 25] )
Z20x20 = AbelianGroup ( [20, 20] )
Z4x100 = AbelianGroup ( [4, 100] )
Z2x10x20 = AbelianGroup ( [2, 10, 20] )
Z505mult = Integers(505).unit_group ()

print( "елементарні дільники " )
print( "Z400: ", Z400.elementary_divisors() )
print( "Z16x25: ", Z16x25.elementary_divisors() )
print( "Z20x20: ", Z20x20.elementary_divisors() )
print( "Z4x100: ", Z4x100.elementary_divisors() )
print( "Z2x10x20: ", Z2x10x20.elementary_divisors() )
print( "Z*505: ", Z505mult.elementary_divisors() )

list_of_groups = [Z400, Z16x25, Z20x20, Z4x100, Z2x10x20, Z505mult]
names = ["Z400", "Z16 x Z25", "Z20 x Z20", "Z4 x Z100", "Z2 x Z10 x Z20", "Z*505"]
list_of_iso_pairs = [];

for A in list_of_groups:
    print(f"група {A}")
    
    p_group = AbelianGroup(A.elementary_divisors())
    print("Розклад на циклічні p-групи:", p_group)
    
    orders_count = {}
    for a in A:
        ord_val = a.order()
        if ord_val in orders_count:
            orders_count[ord_val] += 1
        else:
            orders_count[ord_val] = 1 # словник порядків
    print("Кількість елементів кожного можливого порядку:")
    for order_val in sorted(orders_count.keys()):
        print(f"порядок {order_val}: {orders_count[order_val]}")
    print()

for i in range (len(list_of_groups)):
    for j in range(i+1, len(list_of_groups)):
        A = list_of_groups[i]
        B = list_of_groups[j]
        if A.is_isomorphic(B):
            list_of_iso_pairs.append((names[i], names[j]))
    
print( "всі ізоморфні групи: ")
print( list_of_iso_pairs )
```

1. Реалізуйте функцію, яка для заданого натурального числа $n$ обчислює кількість попарно неізоморфних абелевих груп порядку $n$.
2. Знайдіть перші сто таких натуральних чисел $n$, для яких існує рівно одна абелева група порядку $n$. Для кожного знайденого $n$ проаналізуйте його розклад на прості множники.
3. Сформулюйте гіпотезу: для яких $n\in\mathbb{N}$ існує рівно одна (з точністю до ізоморфізму) абелева група порядку $n$.


```sage
def number_of_abelian(n):
    factors = factor(n)
    k = 1
    for base, power in factors:
        elem = number_of_partitions(power) # шукаємо всі можливі розклади степеня
        k*=elem
    return k

n = int( input("введіть число: ") )
print("Кількість абелевих груп:", number_of_abelian(n))

print("задача 2")
d = 1; # d = n, задача 2
count = 0;
while ( count<100 ):
    if number_of_abelian(d)==1:
        count+=1
        print("number: ", d, "розклад: ", factor(d))
    d+=1
print( count)

# висновок: це такі числа, що або вже є простими, або розкладаються на добуток різних простих в першому степені. для n=1 відповідно існує одна тривіальна абелева група
```

### 4. Факторкільця кільця многочленів
Нехай $f(x)=x^4+x+2$. Виконайте наступні кроки для кожного з кілець $R=\mathbb{Q}, \mathbb{Z}_4, \mathbb{Z}_5$:
1. Побудуйте ідеал $I=(f(x))$ у кільці $R[x]$.
2. Побудуйте факторкільце $R/I$ та знайдіть його порядок.
3. Визначте, чи є факторкільце $R/I$ полем.
4. Знайдіть дільники нуля у факторкільці $R/I$.
5. Розкладіь $f(x)$ на лінійні множники над $R$ та порівняйте з дільниками нуля у факторкільці $R/I$.
6. Перевірте, чи елемент $x+I\in R/I$ є оборотним.

```sage
f = x ^4 + x + 2; # введення многочлену
R1.<x> = QQ[]; # введення кілець
R2.<x> = IntegerModRing(4)[];
R3.<x> = IntegerModRing(5)[];
list_of_rings = [R1, R2, R3]

for i in list_of_rings:
    print(f"ring {i}: ")
    I = i.ideal(f) # побудова ідеалу
    
    Q = i.quotient( I ) # побудова фактор-кільця
    print ( "Фактор-кільце за ідеалом:", Q )
    print ( "Чи є фактор-кільце полем? ", Q.is_field() )
    if Q.is_finite():
        print("Порядок фактор-кільця:", Q.order())
        zero_divs = [a for a in Q if a != 0 and any(a * b == 0 and
                                                    b != 0 for b in Q)]
        print (f"дільники 0: ", zero_divs )
        print("Кількість: ", len( zero_divs ) )
    
        f_in_ring = i(f)  # Многочлен f у поточному кільці i
        x_ring = i.gen()  # Змінна x поточного кільця i
        factors = [x_ring - r for r in i.base_ring() if f_in_ring(r) == 0]
        print("Лінійні множники:", factors)
    else:
        print("Порядок фактор-кільця: нескінченний")
    
    elem = Q(x);
    print( "Чи належить elem фактор-кільцю? ", elem not in zero_divs and 
          elem != 0 )
    
    print("")
```

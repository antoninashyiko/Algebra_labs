### 3. Підгрупи та їхні властивості

Для кожної з груп $\mathbb{Z}_{24}$, $Q_8$, $D_5$, $S_4$ виконайте наступні кроки:
1. Побудуйте групу.
2. Отримайте та виведіть список усіх підгруп групи.   
3. Визначте кількість підгруп групи з точністю до ізоморфізму.
4. Визначте порядки всіх підгруп групи.
5. Перевірте, чи порядок кожної підгрупи ділить порядок групи.

```sage
Z24 = AdditiveAbelianGroup([24]) 
subs = []
for element in Z24.list():
    sub = Z24.submodule([element])
    if sub not in subs:
        subs.append(sub)
k=0
print ("підгрупи Z24:")
for i , sub in enumerate (subs , start =1):
    print (f"підгрупа {i}:")
    print (f"порядок підгрупи: {sub. order ()}")
    if sub.short_name() != Z24.short_name():
        k += 1
    if (Z24.order()%sub.order()==0):
        print("перевірка на подільність порядку підгрупи пройдена")
    print(" ")
print(f"Кількість підгруп, що не ізоморфні Z24: {k}")
```

```sage
Q8=QuaternionGroup ()
subs = Q8.subgroups ()
k=0
print ("підгрупи Q8:")
for i , sub in enumerate (subs , start =1):
    print (f"підгрупа {i}:")
    print (f"порядок підгрупи: {sub. order ()}")
    if (sub.is_isomorphic(Q8)!=1):
        k+=1
    if (Q8.order()%sub.order()==0):
        print("перевірка на подільність порядку підгрупи пройдена")
    print(" ")
print(f"Кількість підгруп, що не ізоморфні Q8: {k}")
```

```sage
D5=DihedralGroup(5)
subs = D5.subgroups ()
k=0
print ("підгрупи D5:")
for i , sub in enumerate (subs , start =1):
    print (f"підгрупа {i}:")
    print (f"порядок підгрупи: {sub. order ()}")
    if (sub.is_isomorphic(D5)!=1):
        k+=1
    if (D5.order()%sub.order()==0):
        print("перевірка на подільність порядку підгрупи пройдена")
    print(" ")
print(f"Кількість підгруп, що не ізоморфні D5: {k}")
```

### 4. Графи Келі

Графом Келі групи $G$ відносно системи твірних $S$ називається орієнтований граф $\Gamma(G,S)$, вершинами якого є елементи групи і кожний $g\in G$ з'єднаний орієнтованим ребро з $gs$ для всіх $s\in S$.

Для кожної з груп: $D_4$, $Q_8$, $S_3$, $S_4$, $A_4$, виконайте наступні кроки:
1. Побудуйте граф Келі для деякої системи твірних групи.
2. Визначте степінь вершин графа Келі.
3. Перевірте, чи граф Келі є вершинно-транзитивним.
4. Дослідіть, чи граф Келі містить гамільтонів цикл.

```sage
D4 = DihedralGroup(4) 
gens = D4.gens () 
CG = D4.cayley_graph(generators=gens)
print("Степені всіх вершин:", CG.degree()) #.in_degree() та .out_degree() - якщо потрібно показати окрмео вхідні та вихідні степені
print("Чи є граф вершинно-транзитивним?:", CG.is_vertex_transitive()) # вершинно-транзитивний граф
print("Чи є Гамільтонів цикл?:", CG.is_hamiltonian())
CG.plot()
```

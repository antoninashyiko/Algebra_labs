### 1. Класи суміжності та нормальні підгрупи

Розглянемо групу $S_4$ та її підгрупи:<br>
$\quad$ (a) $H_1=\langle (1,2,3,4)\rangle$;<br>
$\quad$ (b) $H_2=\langle (1,3), (1,2,3,4)\rangle$;<br>
$\quad$ (c) $H_3=K_4$.

Для кожної підгрупи $H$ виконайте наступні кроки:
1. Обчисліть порядок підгрупи $H$ та її індекс у групі $[S_4 : H]$.
2. Знайдіть ліві та праві класи суміжності підгрупи $H$ у $S_4$.
3. Визначте, чи є $H$ нормальною підгрупою $S_4$.
4. Якщо $H$ є нормальною, то побудуйте факторгрупу $G/H$, її таблицю Келі та визначіть структуру.
5. Якщо $H$ не є нормальною, то знайдіть такий елемент $a\in S_4$, що $aH\neq Ha$.


```sage
S4=SymmetricGroup(4)
H1=S4.subgroup([S4("(1, 2, 3, 4)")])
H2=S4.subgroup([S4("(1, 3)"), S4("(1, 2, 3, 4)")])
H3=S4.subgroup ([ S4 ("(1 ,2) (3 ,4)") , S4 ("(1 ,3) (2 ,4)")])

order1=H1.order(); # порядки підгруп
order2=H2.order();
order3=H3.order();
print("порядок Н1: ", order1)
print("порядок Н2: ", order2)
print("порядок Н3: ", order3)
print("")

index1=S4.order()/order1 # індекси
index2=S4.order()/order2
index3=S4.order()/order3
print("індекс Н1: ", index1)
print("індекс Н2: ", index2)
print("індекс Н3: ", index3)
print("")

subgroup_list=[H1, H2, H3] # пошук нормальних підгруп 
for i in subgroup_list:
    if i.is_normal(S4):
        print(f"підгруа {i} є нормальною")
        F = S4.quotient( i ) # фактор група
        print (f"таблиця Келі для елементу {i}:\n", F. cayley_table () )
    else:
        print(f"підгруа {i} не є нормальною")
        for a in S4 :
            left_coset = [ a*h for h in i]
            right_coset = [h *a for h in i]
        if set( left_coset ) != set( right_coset ):
            print ("знайдено таке а, що a{i} != {i}a: ", a)
            print ("aH =", left_coset )
            print ("Ha =", right_coset )
    print("")
```

### 2. Нормальні підгрупи в діедральних групах

Розгляньте діедральні групи $D_n$ для значень $n\in\{3,4,\dots,30\}$ і виконайте такі кроки:

2. Визначте кількість нормальних підгруп у $D_n$.
3. Складіть перелік порядків усіх нормальних підгруп.

```sage
for n in range(3, 31):
    Dn=DihedralGroup(n) #формуємо дієдральну групу порядку n
    print(f"|||дієдральна група {n}|||")
    k_normal_subgroups=0
    list_order_normal_subgroups=[]
    for H in Dn.subgroups():
        if H.is_normal(Dn):
            k_normal_subgroups+=1 #кількість нормальних підгруп
            list_order_normal_subgroups.append(H.order())
            if k_normal_subgroups<=3: # щоб не виводити всі групи, вивід великий
                print("нормальна підгрупа: ", H) # алгебраїчна структура
                print("порядок: ", H.order())
                print("структура: ", H.structure_description())
                print("таблиця Келі: \n", H.cayley_table())
                print(" ")
    print("загальна кількість нормальний підгруп: ", k_normal_subgroups)
    print(f"список всіх порядків нормальних підгруп D{n}",list_order_normal_subgroups )
    print("")
print("Висновок 5: ")
print("При n - просте, кількість нормальних підгруп 3")
print("При n - непарне, кількість норм. підгруп: кількість дільників + 1 (всі підгрупи поворотів, які відповідають кожному дільнику)")
print("При n - парне, кількість норм. підгруп: кількість дільників + 3 (всі підгрупи поворотів і окремі змішані групи, які відповідаються кожному дільнику)")

# Висновок 6: 
# Про структуру і кількість норм. підгруп описано вище
# Для групи Dn, n=470448
# n - парне, тому буде останній випадок, отже є автоматично 3 нормальні підгрупи: сама група і дві дієдральні n/2(без врахування тривіальної)
# решта рівна кількості дільників 470448 і складається з кутів повороту
# 
# розкласти потрібно на прості(обчилення малі все-таки є): 2^4 * 3^5 * 11^2
# тобто, кількість дільників: 5*6*3=90 і загальна кількість норм. підгруп=93
div=divisors(470448)
print("Кількість норм. підгруп для D470448", len(div)+3)
```

### 3. Гомоморфізм групи матриць у симетричну групу. 

Розглянемо групу $GL_2(\mathbb{Z}_3)$ та множину прямих $M=\{\ell_1, \ell_2, \ell_3, \ell_4\}$ у просторі $\mathbb{Z}_3^2$, що визначаються парами пропорційних ненульових векторів:
\begin{align*}
\ell_1 = \{ (1,0), (2,0) \}, \
\ell_2 = \{ (0,1), (0,2) \}, \
\ell_3 = \{ (1,1), (2,2) \}, \
\ell_4 = \{ (1,2), (2,1) \}.
\end{align*}

1. Побудуйте групу $GL_2(\mathbb{Z}_3)$.
2. Задайте відображення
    $$
    \phi: GL_2(\mathbb{Z}_3) \longrightarrow S_4,
    $$
    яке переводить матрицю у перестановку, що описує її дію на прямі з $M$.
3. Перевірте, що $\phi$ є гомоморфізмом груп.<br>
4. Обчисліть ядро $N=\ker(\phi)$ та образ $\mathrm{Im}(\phi)$.
5. Побудуйте факторгрупу $GL_2(\mathbb{Z}_3)/N$ та порівняйте її з образом $\mathrm{Im}(\phi)$.



```sage
G = GL(2, GF(3));  # група gl2(z3)
S4=SymmetricGroup(4);
l=GF(3)^2
M = {
    1: [l((1, 0)), l((2, 0))],
    2: [l((0, 1)), l((0, 2))],
    3: [l((1, 1)), l((2, 2))],
    4: [l((1, 2)), l((2, 1))]
}
def get_line_index(vector):
    for idx, vectors in M.items():
        if vector in vectors:
            return idx
    return None

list_images=[]
for g in G.gens():
    gmatrix=g.matrix() #з абстрактного в матрицю
    result=[] #список для запису куди яка пряма переходить
    for i in sorted(M.keys()):
        v = M[i][0]
        new_vector = gmatrix * v
        line_image=get_line_index(new_vector)
        result.append(line_image)
    list_images.append(S4(result))
    
phi = G.hom(list_images) 

A = G.random_element() # додаткова перевірка на гомомрфізм
B = G.random_element()
left_side = phi(A * B)
right_side = phi(A) * phi(B)
print(f"phi(A * B) = {left_side}")
print(f"phi(A) * phi(B) = {right_side}")

N=phi.kernel()
print ("Ядро:", N )
print ("Структура ядра:", N.structure_description () )

img = phi.image ( G )
print ("Образ:", img )
print ("Структура образу: ", img . structure_description () )

print("побудова фактор групи для перевірки основної теореми не вийшла, вбудована функція не працює, але оскільки відображення гарантовано гомоморфізм, то факторгрупа по ядру автоматично ізоморфна образу
print("Фактор-група за ядром порядок: ", img.order())
print ("Структура фактор-групи за ядром:", img . structure_description () )
```
